```python
#if condition 
num=int(input("enter the num"))
if num % 3==0:
    print(f'{num} is divisible by 3')
else:
    print(f'{num} is not divisible by 3')
    
```

```python
# check a number is divisible with 3 and 5 or not
num=int(input("enter the num"))
if num % 3==0 or num % 5==0:
    print(f'{num} is divisible by either 3 and 5')
else:
    print(f'{num} is not divisible by either 3 and 5')
```

```python
# check weather the num is even or odd
num=int(input("enter the num"))
if num % 2==0:
        print("even")
else:
        print("odd")
```

```python
# generate total bill if total bill more than 500 then give 10% discount otherwise 0% discount
bill=int(input("enter the bill"))
if bill>500:
    d=bill*(0.1)
    print(f' total bill is 10% discount:{bill-d}')
else:
    print('0% discont')
```

```python
#check weather the product stock is available or not (read product name and stock)
product=input("enter the product")
stock=int(input())
if stock>0:
    print(f"currently {product}:{stock} items available")
else:
     print(f"currently {product}:{stock} out of stock")    
```

```python
# in a shop contains some items along with prices like{'pen':5,'book':'20','pencil:3} a customer want some items from shop,if item is avialabe in shop 
#then find the total amount otherwise 
products={'pen':5,'book':20,'pencil':3}
item=input()
quantity=int(input())
if item in products:# membership operater is used
    total_amount=products.get(item)*quantity# get() syntax is used for read the dict values
    print(f'total amount of {quantity} {item} is : {total_amount}')
else:
    print("product is not available")
```

```python
# check a number is +ve ,-ve or zero
num=int(input())
if num>0:
    print(f'num is positive')
elif num <0:
     print(f'num is negivite')
else:
     print(f'num is zero')
    
```

```python
# divisiable by 3-fizz,5-buzz,both-fizz,buzz
num=int(input())
if num%3==0 and num%5==0:
    print(f'fizz,buzz')
elif num%3==0:
    print(f'fizz')
elif num%3==0:
    print(f'buzz')
else:
    print('none')
```
